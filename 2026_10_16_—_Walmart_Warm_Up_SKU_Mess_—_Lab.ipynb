### Q3 — Build the mapping

A dict from each variant SKU to one canonical SKU. Then apply it in a way that leaves unmapped SKUs as themselves rather than turning them into `NaN` — that single mistake silently deletes most of the catalog.

```python
CANON = {
    'PT-12': 'POPTART-STRAW',
    'SKU#459812': 'POPTART-STRAW',
    'POPTART-STRAW': 'POPTART-STRAW',
    'PT_STRAWBERRY': 'POPTART-STRAW'
}

cat['canonical_sku'] = cat['sku'].map(CANON).fillna(cat['sku'])

print(cat[['sku', 'product_name', 'canonical_sku']])
```

To make a canonical SKU, I made a dictionary called CANON to map the four strawberry Pop-Tart SKUs from question 2 to one canonical SKU, which I labeled POPTART-STRAW. Then, I used .map(CANON) to use the dictionary on the SKU column and make a new column that I called canonical_sku. I also used `.fillna(cat['sku'])` so that any SKUs not included in the dictionary would keep their original values instead of becoming `NaN`, because those products still need to remain in the dataset even if they do not need to be consolidated. Then, I printed the original SKUs, product names, and canonical SKUs to make sure that the mapping worked, and the output shows that the four strawberry Pop-Tart SKUs have the same canonical SKU, and the other products kept their original SKUs. I grouped the four pack and the six pack together as a single product in this, because the lab asks you to group then under a single canonical SKU, but I was unsure if this if different package sizes are supposed to be consolidated.

### Q4 — Normalize `category`

Strip and title-case, then show the distinct values before and after with counts. Note any pair that is still two labels for one thing — `Breakfast` and `Breakfast Foods` are not the same string but they are the same category.

```python
print("Categories before normalization:")
print(cat['category'].value_counts())

cat['category_clean'] = cat['category'].str.strip().str.title()

print("Categories after normalization:")
print(cat['category_clean'].value_counts())
```

I used `.value_counts()` on the `category` column to see the different category names, as well as how many times each showed up before they were normalized. Then, I made the column `category_clean` using `.str.strip()` to get rid of extra spaces and `.str.title()` to make the capitalization consistent. I used `.value_counts()` again to compare the categories after normalization. The output showed that the categories that had different capitalization were combined correctly. For example, Emergency was combined into one category with 9 entries. However, the output also showed that `Breakfast` appeared three times and `Breakfast Foods` appeared once after normalization. The two are referring to the same category, but they are treated as different values because they have different strings. Therefore, in order for the two to not be grouped separately, they would need to be mapped into one category.

### Q5 — Find the fragmentation you have not spotted yet

Hand-scanning a catalog does not scale. **TODO:** build a candidate list programmatically — normalize each product name (lowercase, strip punctuation, sort the words) and group by that, then show any group holding more than one SKU.

This is the workflow from Monday's lecture, and it is how you will actually approach the full catalog on the midterm.

```python
cat['normalized'] = (cat['product_name'].str.lower()
                     .str.replace(r'[^a-z ]', '', regex=True)
                     .str.split().apply(lambda w: ' '.join(sorted(w))))

candidates = (cat.groupby('normalized')['sku']
              .agg(list).loc[lambda s: s.str.len() > 1])

print(candidates)
```

I made a column called `normalized` to find products that might have different SKUs but represent the same item. I used `.str.lower()` to make all product names lowercase and `.str.replace()` to get rid of punctuation and numbers. Then, I used `.str.split()` to separate the product names into words and `sorted()` to put them in alphabetical order. Then, I used `.groupby('normalized')` to group products by their normalized names and `.agg(list)` for their SKUs. I then filtered it to only show groups that had more than one SKU, as those could be representing fragmented products. The output showed that there were 13 groups that could have been duplicated. For example, FLASH-01 and LED-FLASH have different SKUs, but they both have the name "flashlight led" after normalization.
